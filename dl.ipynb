```python
# program 2
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from sklearn.metrics import classification_report

# Load MNIST
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.mnist.load_data()

# Print shapes
print("Shape of X_train:", X_train.shape)
print("Shape of y_train:", y_train.shape)
print("Shape of X_test:", X_test.shape)
print("Shape of y_test:", y_test.shape)

# Display first 10 images
plt.figure(figsize=(12, 6))
for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(X_train[i])
    plt.title(y_train[i])
    plt.axis('off')
plt.show()

# Reshape and normalize
X_train = X_train.reshape(60000, 784).astype("float32") / 255
X_test = X_test.reshape(10000, 784).astype("float32") / 255

print("New shape of X_train:", X_train.shape)
print("New shape of X_test:", X_test.shape)

# Create Deep Feed Forward Neural Network
model = Sequential([
    Dense(128, activation='relu', kernel_initializer='HeNormal', input_shape=(784,)),
    Dense(64, activation='relu', kernel_initializer='HeNormal'),
    Dense(32, activation='relu', kernel_initializer='HeNormal'),
    Dense(10, activation='softmax')
])

# Compile
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# Train
model.fit(X_train, y_train, batch_size=10, epochs=5)

# Predictions
pred_train = np.argmax(model.predict(X_train), axis=1)
pred_test = np.argmax(model.predict(X_test), axis=1)

# Model summary
print("\nModel Summary")
model.summary()

# Evaluation
print("\n--------- Evaluation on Training Data ---------")
print(classification_report(y_train, pred_train))

print("\n--------- Evaluation on Test Data ---------")
print(classification_report(y_test, pred_test))
```

```python
# program 3
import numpy as np
import matplotlib.pyplot as plt
from keras.datasets import fashion_mnist
from keras.models import Sequential
from keras.layers import Conv2D, MaxPooling2D, BatchNormalization
from keras.layers import Activation, Flatten, Dense
from keras.utils import to_categorical

# Load data
(train_X, train_Y), (test_X, test_Y) = fashion_mnist.load_data()

# Reshape and normalize
train_X = train_X.reshape(-1, 28, 28, 1) / 255.0
test_X = test_X.reshape(-1, 28, 28, 1) / 255.0

# One-hot encoding
train_Y = to_categorical(train_Y, 10)
test_Y = to_categorical(test_Y, 10)

# CNN model
model = Sequential()

model.add(Conv2D(256, (3,3), padding='same', input_shape=(28,28,1)))
model.add(BatchNormalization())
model.add(Activation('relu'))
model.add(MaxPooling2D((2,2)))

model.add(BatchNormalization())
model.add(Conv2D(128, (3,3), padding='same'))
model.add(Activation('relu'))
model.add(MaxPooling2D((2,2)))

model.add(BatchNormalization())
model.add(Conv2D(64, (3,3), padding='same'))
model.add(Activation('relu'))
model.add(MaxPooling2D((2,2)))

model.add(Conv2D(28, (3,3), padding='same'))
model.add(Activation('relu'))
model.add(MaxPooling2D((2,2)))

model.add(Flatten())
model.add(Dense(64))
model.add(Dense(10, activation='softmax'))

# Compile
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# Train
model.fit(train_X, train_Y, epochs=5, batch_size=64)

# Test
loss, acc = model.evaluate(test_X, test_Y)
print("Test loss:", loss)
print("Test accuracy:", acc)

# Prediction
predictions = model.predict(test_X)
print(np.argmax(predictions[0]))

# Display image
plt.imshow(test_X[0].reshape(28,28), cmap='gray')
plt.show()
```

```python
# program 4.1
import numpy as np
import matplotlib.pyplot as plt
from keras.datasets import fashion_mnist
from keras.models import Sequential
from keras.layers import Conv2D, MaxPooling2D, Flatten, Dense
from keras.utils import to_categorical

# Load data
(train_X, train_Y), (test_X, test_Y) = fashion_mnist.load_data()

train_X = train_X.reshape(-1, 28, 28, 1).astype('float32') / 255
test_X = test_X.reshape(-1, 28, 28, 1).astype('float32') / 255

train_Y = to_categorical(train_Y, 10)
test_Y = to_categorical(test_Y, 10)

# CNN
model = Sequential([
    Conv2D(32, (3,3), activation='relu', padding='same',
           input_shape=(28,28,1)),
    MaxPooling2D((2,2)),

    Conv2D(64, (3,3), activation='relu', padding='same'),
    MaxPooling2D((2,2)),

    Conv2D(128, (3,3), activation='relu', padding='same'),
    Conv2D(128, (3,3), activation='relu', padding='same'),

    Flatten(),
    Dense(128, activation='relu'),
    Dense(10, activation='softmax')
])

model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

model.fit(
    train_X, train_Y,
    epochs=10,
    batch_size=64,
    validation_split=0.1
)

loss, acc = model.evaluate(test_X, test_Y)
print("Test Loss:", loss)
print("Test Accuracy:", acc)

prediction = model.predict(test_X)
plt.imshow(test_X[0].reshape(28,28), cmap='gray')
plt.title(f"Predicted: {np.argmax(prediction[0])}")
plt.axis('off')
plt.show()
```

```python
# program 4.2
import numpy as np
import matplotlib.pyplot as plt
from keras.datasets import fashion_mnist
from keras.models import Sequential
from keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout
from keras.utils import to_categorical
from keras import regularizers

(train_X, train_Y), (test_X, test_Y) = fashion_mnist.load_data()

train_X = train_X.reshape(-1,28,28,1).astype('float32') / 255
test_X = test_X.reshape(-1,28,28,1).astype('float32') / 255

train_Y = to_categorical(train_Y, 10)
test_Y = to_categorical(test_Y, 10)

l1 = regularizers.l1(0.001)

model = Sequential([
    Conv2D(32, (3,3), padding='same',
           kernel_regularizer=l1, activation='relu',
           input_shape=(28,28,1)),
    MaxPooling2D((2,2)),

    Conv2D(64, (3,3), padding='same',
           kernel_regularizer=l1, activation='relu'),
    MaxPooling2D((2,2)),

    Conv2D(128, (3,3), padding='same',
           kernel_regularizer=l1, activation='relu'),

    Conv2D(128, (3,3), padding='same',
           kernel_regularizer=l1, activation='relu'),

    Flatten(),
    Dense(128, activation='relu', kernel_regularizer=l1),
    Dropout(0.5),
    Dense(10, activation='softmax')
])

model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

model.fit(
    train_X, train_Y,
    epochs=10,
    batch_size=64,
    validation_split=0.1
)

loss, acc = model.evaluate(test_X, test_Y)
print("Test Loss:", loss)
print("Test Accuracy:", acc)

prediction = model.predict(test_X)
plt.imshow(test_X[0].reshape(28,28), cmap='gray')
plt.title(f"Predicted: {np.argmax(prediction[0])}")
plt.axis('off')
plt.show()
```

```python
# program 4.5
import numpy as np
import matplotlib.pyplot as plt
from keras.datasets import fashion_mnist
from keras.models import Sequential
from keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout
from keras.utils import to_categorical
from keras import regularizers

(train_X, train_Y), (test_X, test_Y) = fashion_mnist.load_data()

train_X = train_X.reshape(-1,28,28,1).astype('float32') / 255
test_X = test_X.reshape(-1,28,28,1).astype('float32') / 255

train_Y = to_categorical(train_Y, 10)
test_Y = to_categorical(test_Y, 10)

l2 = regularizers.l2(0.001)

model = Sequential([
    Conv2D(32, (3,3), padding='same',
           kernel_regularizer=l2,
           activation='relu',
           input_shape=(28,28,1)),
    MaxPooling2D((2,2)),

    Conv2D(64, (3,3), padding='same',
           kernel_regularizer=l2,
           activation='relu'),
    MaxPooling2D((2,2)),

    Conv2D(128, (3,3), padding='same',
           kernel_regularizer=l2,
           activation='relu'),

    Conv2D(128, (3,3), padding='same',
           kernel_regularizer=l2,
           activation='relu'),

    Flatten(),
    Dense(128, activation='relu', kernel_regularizer=l2),
    Dropout(0.5),
    Dense(10, activation='softmax')
])

model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

model.fit(
    train_X, train_Y,
    epochs=10,
    batch_size=64,
    validation_split=0.1
)

loss, acc = model.evaluate(test_X, test_Y)
print("Test Loss:", loss)
print("Test Accuracy:", acc)

prediction = model.predict(test_X)

labels = [
    'T-shirt', 'Trouser', 'Pullover', 'Dress', 'Coat',
    'Sandal', 'Shirt', 'Sneaker', 'Bag', 'Ankle boot'
]

predicted_class = np.argmax(prediction[0])
print("Predicted label:", labels[predicted_class])

plt.imshow(test_X[0].reshape(28,28), cmap='gray')
plt.title(f"Predicted: {labels[predicted_class]}")
plt.axis('off')
plt.show()
```

```python
# program 5

import torch
import requests
import matplotlib.pyplot as plt
from PIL import Image
from transformers import BlipProcessor, BlipForConditionalGeneration

# Image
url = "https://storage.googleapis.com/sfr-vision-language-research/BLIP/demo.jpg"
image = Image.open(requests.get(url, stream=True).raw).convert("RGB")

# Display image
plt.imshow(image)
plt.axis("off")
plt.title("Input Image")
plt.show()

# Load model
processor = BlipProcessor.from_pretrained(
    "Salesforce/blip-image-captioning-base"
)
model = BlipForConditionalGeneration.from_pretrained(
    "Salesforce/blip-image-captioning-base"
)

# Prepare image
inputs = processor(image, return_tensors="pt")

# Generate caption
with torch.no_grad():
    output = model.generate(**inputs)

caption = processor.decode(output[0], skip_special_tokens=True)

print("Generated Caption:")
print(caption)
```
